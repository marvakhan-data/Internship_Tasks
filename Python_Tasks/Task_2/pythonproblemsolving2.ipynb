**Example input:**

**Expected output:**

**b)Understand:**

Finding first word that occurs exactly once in a sentence and treating uppercase and lowercase versions as the same word.

**c)Break:**

1. Take the sentence as input.
2. Convert it to lowercase and split it into words.
3. Create an empty dictionary to store the count of each word.
4. Go through each word and add 1 to its count.
5. Go through the dictionary in order and find the first word with a count of 1.
6. Print it.

**d)Logic:**

1. Take the sentence, convert it to lowercase, and split it into words.
2. Count how many times each word appears using a dictionary.
3. Set result to None, meaning nothing found yet.
4. Check each word in the dictionary, in order.
5. If its count is exactly 1, store it in result and stop.
6. A dictionary keeps words in the order they first appeared, so the first match is the first unique word.

```python
#e)Code:
sentence = "apple banana apple mango banana orange"

words = sentence.lower().split()

counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1

result = None
for word in counts:
    if counts[word] == 1:
        result = word
        break

print(result)
```

**Q20. Find the Longest Word That Appears Only Once**

Write a Python program to find the longest word that appears exactly once in a sentence. If
multiple qualifying words have the same maximum length, return the first one encountered.
Treat uppercase and lowercase versions as the same word.

**a)Observe:**

**Example input:**

Python makes coding useful Python

**Expected output:**

coding

**b)Understand:**


Finding  the longest word that appears exactly once in a sentence. If
multiple qualifying words have the same maximum length, return the first one encountered.
Treating uppercase and lowercase versions as the same word.

**c)Break:**

1. Take the sentence as input.
2. Convert it to lowercase and split it into words.
3. Create an empty dictionary to store the count of each word.
4. Go through each word and add 1 to its count.
5. Go through the dictionary and look only at words with a count of 1.
6. Keep track of the longest one found so far.
7. Print it.

**d)Logic:**

1. Take the sentence, convert it to lowercase, and split it into words.
2. Count how many times each word appears using a dictionary.
3. Set result to None, meaning nothing found yet.
4. Check each word in the dictionary, in order.
5. If the word appears exactly once, and result is empty or the word is longer than result, update result.
6. A word of the same length does not replace result, so the first one stays.
7. Print result.

```python
#e)Code:
sentence = "Python makes coding useful Python"

words = sentence.lower().split()

counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1

result = None
for word in counts:
    if counts[word] == 1:
        if result is None or len(word) > len(result):
            result = word

print(result)
```
