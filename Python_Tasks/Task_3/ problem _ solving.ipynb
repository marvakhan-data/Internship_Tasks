Each cell below has the pseudocode, the expected output, the Python code and one extra test case.

```python
# Q1. Word Frequency
# Pseudocode:
#   1. Make an empty dictionary
#   2. Split the sentence into words
#   3. For each word, add 1 to its count (start from 0 if it is new)
#   4. Return the dictionary
# Input: "python is easy and python is powerful"
# Expected output: {'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}

def word_frequency(sentence):
    freq = {}
    for word in sentence.split():
        freq[word] = freq.get(word, 0) + 1
    return freq

print(word_frequency("python is easy and python is powerful"))
print(word_frequency("to be or not to be"))  # extra test -> {'to': 2, 'be': 2, 'or': 1, 'not': 1}
```

```python
# Q2. Character Frequency
# Pseudocode:
#   1. Make an empty dictionary
#   2. Go through each character in the word
#   3. Add 1 to that character's count
#   4. Return the dictionary
# Input: "banana"
# Expected output: {'b': 1, 'a': 3, 'n': 2}

def char_frequency(word):
    freq = {}
    for ch in word:
        freq[ch] = freq.get(ch, 0) + 1
    return freq

print(char_frequency("banana"))
print(char_frequency("hello"))  # extra test -> {'h': 1, 'e': 1, 'l': 2, 'o': 1}
```

```python
# Q3. First Non-Repeating Character
# Pseudocode:
#   1. Count how many times each character appears
#   2. Go through the word from left to right
#   3. Return the first character whose count is 1
#   4. If there is none, return None
# Input: "swiss"
# Expected output: w

def first_non_repeating(word):
    freq = {}
    for ch in word:
        freq[ch] = freq.get(ch, 0) + 1
    for ch in word:
        if freq[ch] == 1:
            return ch
    return None

print(first_non_repeating("swiss"))
print(first_non_repeating("aabbc"))  # extra test -> c
```

```python
# Q4. Anagram Check
# Pseudocode:
#   1. Count the characters of the first string
#   2. Count the characters of the second string
#   3. If both counts are equal, they are anagrams, otherwise not
# Input: s1 = "listen", s2 = "silent"
# Expected output: True

def are_anagrams(s1, s2):
    def count(s):
        d = {}
        for ch in s:
            d[ch] = d.get(ch, 0) + 1
        return d
    return count(s1) == count(s2)

print(are_anagrams("listen", "silent"))
print(are_anagrams("hello", "world"))  # extra test -> False
```

```python
# Q5. First Repeating Character
# Pseudocode:
#   1. Keep an empty set of characters seen so far
#   2. Read the string from left to right
#   3. If the character is already in the set, return it
#   4. Otherwise add it to the set
#   5. If the loop finishes, return None
# Input: "programming"
# Expected output: r

def first_repeating(text):
    seen = set()
    for ch in text:
        if ch in seen:
            return ch
        seen.add(ch)
    return None

print(first_repeating("programming"))
print(first_repeating("abc"))  # extra test -> None
```

```python
# Q6. Move Zeros to End
# Pseudocode:
#   1. Pick out all non-zero numbers, keeping their order
#   2. Count how many zeros were left out
#   3. Add that many zeros to the end
#   4. Return the new list
# Input: [0, 3, 0, 1, 2, 0, 4]
# Expected output: [3, 1, 2, 4, 0, 0, 0]

def move_zeros_to_end(nums):
    non_zeros = [n for n in nums if n != 0]
    zero_count = len(nums) - len(non_zeros)
    return non_zeros + [0] * zero_count

print(move_zeros_to_end([0, 3, 0, 1, 2, 0, 4]))
print(move_zeros_to_end([0, 0, 5, 0, 9]))  # extra test -> [5, 9, 0, 0, 0]
```

```python
# Q7. Second Highest Distinct Number
# Pseudocode:
#   1. Remove duplicates by converting the list to a set
#   2. Sort the unique numbers from largest to smallest
#   3. If there are fewer than 2 numbers, return None
#   4. Otherwise return the second number
# Input: [10, 20, 40, 30, 40, 20]
# Expected output: 30

def second_highest(nums):
    distinct = sorted(set(nums), reverse=True)
    if len(distinct) < 2:
        return None
    return distinct[1]

print(second_highest([10, 20, 40, 30, 40, 20]))
print(second_highest([5, 5, 5]))  # extra test -> None
```

```python
# Q8. Remove Duplicates (Keep Order)
# Pseudocode:
#   1. Make an empty set (seen) and an empty list (result)
#   2. For each number: if it is not in seen, add it to both seen and result
#   3. If it is already in seen, skip it
#   4. Return result
# Input: [4, 2, 4, 5, 2, 7, 5]
# Expected output: [4, 2, 5, 7]

def remove_duplicates(nums):
    seen = set()
    result = []
    for n in nums:
        if n not in seen:
            seen.add(n)
            result.append(n)
    return result

print(remove_duplicates([4, 2, 4, 5, 2, 7, 5]))
print(remove_duplicates([1, 1, 1, 2, 3, 3]))  # extra test -> [1, 2, 3]
```

```python
# Q9. Missing Number from 1 to N
# Pseudocode:
#   1. Work out what the sum of 1 to N should be: N * (N + 1) / 2
#   2. Add up the numbers in the list
#   3. Expected sum minus actual sum is the missing number
# Input: nums = [1, 2, 4, 5, 6], N = 6
# Expected output: 3

def missing_number(nums, n):
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)

print(missing_number([1, 2, 4, 5, 6], 6))
print(missing_number([2, 3, 1, 5], 5))  # extra test -> 4
```

```python
# Q10. Two Numbers Equal to Target
# Pseudocode:
#   1. Keep a set of numbers already seen
#   2. For each number, find what it needs: target - number
#   3. If that needed number was already seen, return [needed, number]
#   4. Otherwise add the number to seen
#   5. If no pair is found, return an empty list
# Input: nums = [2, 7, 11, 15], target = 9
# Expected output: [2, 7]

def two_sum_pair(nums, target):
    seen = set()
    for num in nums:
        complement = target - num
        if complement in seen:
            return [complement, num]
        seen.add(num)
    return []

print(two_sum_pair([2, 7, 11, 15], 9))
print(two_sum_pair([1, 2, 3], 10))  # extra test -> []
```

```python
# Q11. Number Closest to Zero
# Pseudocode:
#   1. If the list is empty, return None
#   2. Compare numbers by their distance from zero (absolute value)
#   3. If two numbers are equally close, choose the positive one
#   4. Return the winner
# Input: [-8, -3, 2, 5, -2]
# Expected output: 2

def closest_to_zero(nums):
    if not nums:
        return None
    return min(nums, key=lambda x: (abs(x), -x))

print(closest_to_zero([-8, -3, 2, 5, -2]))
print(closest_to_zero([-4, 4, 7]))  # extra test -> 4
```

```python
# Q12. Common Elements
# Pseudocode:
#   1. Convert list2 to a set so lookups are quick
#   2. Go through list1 in order
#   3. If the item is in list2 and not already in the result, add it
#   4. Return the result
# Input: list1 = [1, 2, 3, 4, 5], list2 = [3, 4, 5, 6, 7]
# Expected output: [3, 4, 5]

def common_elements(list1, list2):
    set2 = set(list2)
    added = set()
    result = []
    for item in list1:
        if item in set2 and item not in added:
            result.append(item)
            added.add(item)
    return result

print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))
print(common_elements([1, 2, 2, 3], [2, 3, 3]))  # extra test -> [2, 3]
```

```python
# Q13. Rotate Right by K
# Pseudocode:
#   1. If the list is empty, return an empty list
#   2. Use k % length so a big K still works
#   3. Take the last k items and put them in front of the remaining items
#   4. Return the new list
# Input: nums = [1, 2, 3, 4, 5], K = 2
# Expected output: [4, 5, 1, 2, 3]

def rotate_right(nums, k):
    if not nums:
        return []
    k = k % len(nums)
    if k == 0:
        return nums[:]
    return nums[-k:] + nums[:-k]

print(rotate_right([1, 2, 3, 4, 5], 2))
print(rotate_right([1, 2, 3], 4))  # extra test -> [3, 1, 2]
```

```python
# Q14. Max Sum of Two Consecutive Elements
# Pseudocode:
#   1. Set best = first element + second element
#   2. Move through the list, adding each element to the one next to it
#   3. If that sum is bigger than best, update best
#   4. Return best
# Input: [4, 7, 2, 9, 5]
# Expected output: 14

def max_consecutive_sum(nums):
    best = nums[0] + nums[1]
    for i in range(1, len(nums) - 1):
        best = max(best, nums[i] + nums[i + 1])
    return best

print(max_consecutive_sum([4, 7, 2, 9, 5]))
print(max_consecutive_sum([-1, -5, -2]))  # extra test -> -6
```

```python
# Q15. All Pairs Equal to Target
# Pseudocode:
#   1. Count how many times each number appears
#   2. For each number a, work out b = target - a
#   3. If a < b and b exists, save the pair [a, b] (a < b avoids duplicate pairs)
#   4. If a == b, save it only when the number appears at least twice
#   5. Sort the pairs by their first number and return them
# Input: nums = [2, 4, 3, 5, 7, 8, 1], target = 9
# Expected output: [[1, 8], [2, 7], [4, 5]]

def all_pairs_with_sum(nums, target):
    counts = {}
    for n in nums:
        counts[n] = counts.get(n, 0) + 1

    pairs = []
    for a in counts:
        b = target - a
        if a < b and b in counts:
            pairs.append([a, b])
        elif a == b and counts[a] >= 2:
            pairs.append([a, b])
    pairs.sort(key=lambda p: p[0])
    return pairs

print(all_pairs_with_sum([2, 4, 3, 5, 7, 8, 1], 9))
print(all_pairs_with_sum([3, 3, 3, 6], 6))  # extra test -> [[3, 3]]
```
