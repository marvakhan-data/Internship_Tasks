```python
# Q1
# Prediction:
# <class 'dict'>

x = dict()
print(type(x))

# Explanation: dict() with no arguments builds an empty dictionary, so type() reports the dict class.
```

```python
# Q2
# Prediction:
# 0
# <class 'set'>

x = set()
print(len(x))
print(type(x))

# Explanation: set() is the only way to make an empty set. It holds nothing (length 0) and its type is set.
```

```python
# Q3
# Prediction:
# False

x = {}
print(bool(x))

# Explanation: {} is an empty dict, not a set. Empty containers are falsy, so bool() gives False.
```

```python
# Q4
# Prediction:
# {10}

x = set()
x.add(10)
print(x)

# Explanation: x really is a set, so add() inserts 10 and the set prints with one element.
```

```python
# Q5
# Prediction:
# AttributeError: 'dict' object has no attribute 'add'

x = {}
x.add(10)
print(x)

# Explanation: {} creates a dict, not a set (the empty set trap). Dicts have no add() method, so Python raises AttributeError.
```

```python
# Q6
# Prediction:
# <class 'int'>

x = (25)
print(type(x))

# Explanation: Parentheses alone only group an expression. Without a comma, (25) is just the integer 25.
```

```python
# Q7
# Prediction:
# 1

x = (25,)
print(len(x))

# Explanation: The trailing comma is what makes a tuple. This is a one-element tuple, so len is 1.
```

```python
# Q8
# Prediction:
# 25

x = (25,)
print(x[0])

# Explanation: Tuples are indexed like lists. Index 0 is the only element, 25.
```

```python
# Q9
# Prediction:
# <class 'tuple'>

x = 25, 30
print(type(x))

# Explanation: Commas create the tuple (tuple packing); parentheses are optional.
```

```python
# Q10
# Prediction:
# 35

x = (25, 30, 35)
print(x[-1])

# Explanation: Negative index -1 counts from the end, so it picks the last element.
```

```python
# Q11
# Prediction:
# False
# True

print(bool(0))
print(bool(10))

# Explanation: Zero is the falsy number; every non-zero number is truthy.
```

```python
# Q12
# Prediction:
# False
# True

print(bool(""))
print(bool(" "))

# Explanation: An empty string is falsy, but a string containing a space has length 1, so it is truthy.
```

```python
# Q13
# Prediction:
# False
# True

print(bool([]))
print(bool([] + [1]))

# Explanation: [] is empty, hence falsy. [] + [1] concatenates to [1], a non-empty list, hence truthy.
```

```python
# Q14
# Prediction:
# False
# True

print(bool(()))
print(bool((1,)))

# Explanation: An empty tuple is falsy; a tuple with one element is truthy.
```

```python
# Q15
# Prediction:
# False
# True

print(bool(None))
print(bool("None"))

# Explanation: None is the null object and is falsy. The string "None" is just 4 characters of text, so it is truthy.
```

```python
# Q16
# Prediction:
# 1
# {True: 'B'}

d = {True: "A", 1: "B"}
print(len(d))
print(d)

# Explanation: True == 1 and hash(True) == hash(1), so they are the same key. The key stays True (first inserted) but the value is overwritten with 'B'.
```

```python
# Q17
# Prediction:
# B

d = {1: "A", True: "B"}
print(d[True])

# Explanation: 1 and True are the same key. The key stays 1, the value becomes 'B', and d[True] finds it.
```

```python
# Q18
# Prediction:
# {1.0: 'B'}

d = {1.0: "A", True: "B"}
print(d)

# Explanation: 1.0 == True and their hashes match, so one key exists. The original key 1.0 is kept; the value is updated to 'B'.
```

```python
# Q19
# Prediction:
# B

d = {False: "A", 0: "B"}
print(d[0])

# Explanation: False equals 0 with the same hash, so they collide into one key whose value was overwritten by 'B'.
```

```python
# Q20
# Prediction:
# True
# False

a = [10, 20]
b = [10, 20]
print(a == b)
print(a is b)

# Explanation: == compares contents (equal). is compares identity; two separate list literals are two different objects.
```

```python
# Q22
# Prediction:
# True
# False

a = "hello"
b = "".join(["he", "llo"])
print(a == b)
print(a is b)

# Explanation: The text is equal. join() builds a new string at runtime, so it is a different object from the literal (identity result is CPython behaviour, not guaranteed).
```

```python
# Q23
# Prediction:
# True
# False

a = [1]
b = a[:]
print(a == b)
print(a is b)

# Explanation: A full slice makes a new list with the same contents: equal but not identical.
```

```python
# Q24
# Prediction:
# True
# False

a = int("500")
b = int("500")
print(a == b)
print(a is b)

# Explanation: Values are equal. CPython only caches small ints (-5 to 256); 500 created at runtime gives two separate objects.
```

```python
# Q25
# Prediction:
# [10, 20, 30]

a = [10, 20]
b = a
b.append(30)
print(a)

# Explanation: Both names point to one list. append mutates it in place, so a sees the change.
```

```python
# Q26
# Prediction:
# [10, 20]

a = [10, 20]
b = a
a = a + [30]
print(b)

# Explanation: + builds a new list and rebinds a to it. b still points to the original, unchanged list.
```

```python
# Q27
# Prediction:
# [10, 20, 30]

a = [10, 20]
b = a
a += [30]
print(b)

# Explanation: += on a list extends it in place (like extend), so the shared object changes and b sees it.
```

```python
# Q28
# Prediction:
# [20, 30]

a = [10, 20, 30, 40]
b = a[1:3]
print(b)

# Explanation: Slice start is inclusive and stop is exclusive: indexes 1 and 2.
```

```python
# Q29
# Prediction:
# [10, 30]

a = [10, 20, 30, 40]
b = a[::2]
print(b)

# Explanation: Step 2 takes every second element starting at index 0.
```

```python
# Q31
# Prediction:
# [1, 2, 3]

a = [1, 2, 3]
b = a
del a
print(b)

# Explanation: del a only removes the name a. The list object survives because b still references it.
```

```python
# Q32
# Prediction:
# [2, 3]

a = [1, 2, 3]
b = a
del a[0]
print(b)

# Explanation: del a[0] deletes an element from the shared object itself, so b shows the change.
```

```python
# Q33
# Prediction:
# []

a = [1, 2, 3]
b = a
del a[:]
print(b)

# Explanation: del a[:] removes all elements in place from the shared list, leaving it empty.
```

```python
# Q34
# Prediction:
# NameError: name 'a' is not defined

a = [1, 2, 3]
del a
print(a)

# Explanation: del a removes the name from the namespace, so using a afterwards raises NameError.
```

```python
# Q35
# Prediction:
# [[1, 99], [3, 4]]

import copy
a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)

# Explanation: A shallow copy makes a new outer list but shares the inner lists, so editing an inner list shows up in a.
```

```python
# Q36
# Prediction:
# [[1, 2], [3, 4]]

import copy
a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)

# Explanation: deepcopy duplicates the inner lists too, so a is untouched.
```

```python
# Q37
# Prediction:
# [[10], [20, 30]]

import copy
a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)

# Explanation: list.copy() is shallow; the inner list [20] is shared, so appending through a appears in b.
```

```python
# Q38
# Prediction:
# [[10], [20]]

import copy
a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)

# Explanation: deepcopy gave b its own inner lists, independent of a.
```

```python
# Q39
# Prediction:
# False
# True

import copy
a = [[1, 2], [3, 4]]
b = copy.copy(a)
print(a is b)
print(a[0] is b[0])

# Explanation: The outer lists are different objects, but a shallow copy reuses the same inner list objects.
```
